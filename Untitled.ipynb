```python
import torch

print(torch.cuda.is_available())
```

```python
from sklearn.datasets import fetch_openml

mnist = fetch_openml('mnist_784', as_frame=False)
```

```python
X , y = mnist.data , mnist.target
```

```python
print(X.shape , y.shape)
```

```python
import matplotlib.pyplot as plt 

def plot_digit(image_data):
    image = image_data.reshape(28,28)
    plt.imshow(image ,cmap='binary')
    plt.axis('off')
```

```python
some_digit = X[0]
plot_digit(some_digit)
plt.show()
```

```python
plt.figure(figsize=(9,9))
for idx , image in enumerate(X[:100]):
    plt.subplot(10 , 10 , idx+1)
    plot_digit(image)
plt.show()
```

```python
X_train , X_test , y_train , y_test = X[:60000] , X[60000:] , y[:60000] , y[60000:]
```

```python
# Binary classification 

y_train_5 = (y_train == '5')
y_test_5 = (y_test == '5')
```

```python
print(f"This is training data : {y_train_5}   \n This is testing data :  {y_test_5}")
```

```python
from sklearn.linear_model import SGDClassifier 

sgd_clf = SGDClassifier(random_state=42)
sgd_clf.fit(X_train , y_train_5)
```

```python
#Getting the Performance of the model using cross_validation_score

from sklearn.model_selection import cross_val_score

cross_val_score(sgd_clf , X_train , y_train_5 , cv=3 , scoring="accuracy")
```

```python
# CLoning the model using clone function and then fitting it again using skfolds 
from sklearn.model_selection import StratifiedKFold
from sklearn.base import clone 

skfolds = StratifiedKFold(n_splits=3)  # add shuffle=True if the dataset is not
                                       # already shuffled
for train_index, test_index in skfolds.split(X_train, y_train_5):
    clone_clf = clone(sgd_clf)
    X_train_folds = X_train[train_index]
    y_train_folds = y_train_5[train_index]
    X_test_fold = X_train[test_index]
    y_test_fold = y_train_5[test_index]

    clone_clf.fit(X_train_folds, y_train_folds)
    y_pred = clone_clf.predict(X_test_fold)
    n_correct = sum(y_pred == y_test_fold)
    print(n_correct / len(y_pred))
```

```python
from sklearn.model_selection import cross_val_predict

y_train_predict = cross_val_predict(sgd_clf , X_train , y_train_5 , cv=3)
```

```python
# Using Confusion matrix to get more information about that 
from sklearn.metrics import confusion_matrix 
cm = confusion_matrix(y_train_5 , y_train_predict)
print(cm)
```
